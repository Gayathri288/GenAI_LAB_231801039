```python
from collections import defaultdict, Counter
import nltk
from nltk.corpus import brown
```

```python
nltk.download('brown')
```

```python
sentences = brown.sents()
print("Total Sentences:", len(sentences))
```

```python
tokens = [word.lower() for sent in sentences for word in sent]
print("Total Tokens:", len(tokens))
```

```python
def build_ngram(tokens, n):
    model = defaultdict(Counter)
    for i in range(len(tokens) - n):
        context = tuple(tokens[i:i+n-1])
        next_word = tokens[i+n-1]
        model[context][next_word] += 1
    return model
```

```python
bigram_model = build_ngram(tokens, 2)
print("Bigram Model Built Successfully")
```

```python
trigram_model = build_ngram(tokens, 3)
print("Trigram Model Built Successfully")
```

```python
def predict_next(context, model):
    context = tuple(context.lower().split())
    if context not in model:
        return "No prediction found"
    return model[context].most_common(5)
```

```python
print("Bigram Prediction for 'government':")
print(predict_next("government", bigram_model))
```

```python
print("Trigram Prediction for 'the government':")
print(predict_next("the government", trigram_model))
```
